### Вариант задания №14


<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Описание задачи:
Создать базовый класс Supplier в C#, который будет представлять информацию о
поставщиках товаров или услуг. На основе этого класса разработать 2-3
производных класса, демонстрирующих принципы наследования и полиморфизма.
В каждом из классов должны быть реализованы новые атрибуты и методы, а также
переопределены некоторые методы базового класса для демонстрации
полиморфизма.
Требования к базовому классу Supplier:
• Атрибуты: ID поставщика (SupplierId), Название компании (CompanyName),
Тип продукции (ProductType).
• Методы:
o GetCompanyInfo(): метод для получения информации о компании.
o ProvideQuote(): метод для предоставления котировки на товары или
услуги.
o SubmitOrder(): метод для отправки заказа поставщику.
Требования к производным классам:
1. Производитель (Manufacturer): Должен содержать дополнительные атрибуты,
такие как Год основания (FoundedYear). Метод ProvideQuote() должен быть
переопределен для включения информации о годе основания компании в
котировку.
2. Ритейлер (Retailer): Должен содержать дополнительные атрибуты, такие как
Расположение магазина (StoreLocation). Метод SubmitOrder() должен быть
переопределен для добавления информации о расположении магазина при
отправке заказа.
3. Импортер (Importer) (если требуется третий класс): Должен содержать
дополнительные атрибуты, такие как Страна происхождения товара
(OriginCountry). Метод GetCompanyInfo() должен быть переопределен для
отображения страны происхождения товара вместе с остальной
информацией о компании.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте простое, сложное и множественное наследование


<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;
using System.Collections.Generic;

// --- ИНТЕРФЕЙСЫ ДЛЯ МНОЖЕСТВЕННОГО НАСЛЕДОВАНИЯ ---
public interface ICustomsClearance
{
    string DeclarationNumber { get; set; }
    bool ClearCustoms();
}

public interface ILogistics
{
    string TransportType { get; set; }
    void CalculateDeliveryTime(double distanceKm);
}

// --- ВСПОМОГАТЕЛЬНЫЕ КЛАССЫ ---
public class Product
{
    public string Name { get; set; }
    public decimal Price { get; set; }

    public Product(string name, decimal price)
    {
        Name = string.IsNullOrWhiteSpace(name) ? "Неизвестный товар" : name;
        Price = price < 0 ? 0 : price;
    }
}


public class Supplier
{
    // 4 Атрибута (Свойства)
    public int SupplierId { get; set; }
    public string CompanyName { get; set; }
    public string ProductType { get; set; }
    public double Rating { get; set; }

    public Supplier(int supplierId, string companyName, string productType, double rating = 5.0)
    {
        SupplierId = supplierId < 0 ? 0 : supplierId;
        CompanyName = string.IsNullOrWhiteSpace(companyName) ? "Без названия" : companyName;
        ProductType = string.IsNullOrWhiteSpace(productType) ? "Разное" : productType;
        Rating = Math.Clamp(rating, 0.0, 5.0);
    }

    // 4 Метода
    public virtual string GetCompanyInfo()
    {
        return $"[ID: {SupplierId}] {CompanyName} | Продукция: {ProductType} | Рейтинг: {Rating}/5.0";
    }

    public virtual void ProvideQuote(Product product)
    {
        Console.WriteLine($"Котировка от {CompanyName}: {product.Name} по цене {product.Price:C}.");
    }

    public virtual void SubmitOrder(Product product, int quantity)
    {
        Console.WriteLine($"Заказ на {quantity} шт. '{product.Name}' отправлен поставщику {CompanyName}.");
    }

    public void UpdateRating(double newRating)
    {
        Rating = Math.Clamp(newRating, 0.0, 5.0);
        Console.WriteLine($"Рейтинг компании {CompanyName} обновлен до {Rating}.");
    }
}


public class Manufacturer : Supplier
{
    // 4 Дополнительных атрибута
    public int FoundedYear { get; set; }
    public int ProductionCapacity { get; set; }
    public bool HasISO39001Certificate { get; set; }
    public decimal WholesaleDiscount { get; set; }

    public Manufacturer(int supplierId, string companyName, string productType, double rating, 
                        int foundedYear, int productionCapacity, bool hasCert, decimal discount)
        : base(supplierId, companyName, productType, rating)
    {
        FoundedYear = foundedYear;
        ProductionCapacity = productionCapacity;
        HasISO39001Certificate = hasCert;
        WholesaleDiscount = discount;
    }

    // 4 Дополнительных/переопределённых метода
    public override void ProvideQuote(Product product)
    {
        decimal wholesalePrice = product.Price * (1.0m - WholesaleDiscount);
        Console.WriteLine($"Завод {CompanyName} (осн. {FoundedYear} г., скидка {WholesaleDiscount:P0}): Оптовая цена на '{product.Name}' — {wholesalePrice:C}.");
    }

    public void PlanProductionRun(int units)
    {
        Console.WriteLine($"Производство {CompanyName} запустило партию на {units} ед. (Макс. мощность: {ProductionCapacity} ед./мес).");
    }

    public void InspectQuality()
    {
        Console.WriteLine($"Проведена проверка качества на заводе {CompanyName}. Сертификат ISO: {(HasISO39001Certificate ? "Да" : "Нет")}.");
    }

    public override string GetCompanyInfo()
    {
        return $"{base.GetCompanyInfo()} | Осн.: {FoundedYear} г. | Мощность: {ProductionCapacity} ед./мес";
    }
}


public class Retailer : Supplier
{
    // 4 Дополнительных атрибута
    public string StoreLocation { get; set; }
    public int RetailOutletsCount { get; set; }
    public bool HasOnlineShop { get; set; }
    public decimal MarkupPercentage { get; set; }

    public Retailer(int supplierId, string companyName, string productType, double rating,
                    string storeLocation, int outlets, bool onlineShop, decimal markup)
        : base(supplierId, companyName, productType, rating)
    {
        StoreLocation = storeLocation;
        RetailOutletsCount = outlets;
        HasOnlineShop = onlineShop;
        MarkupPercentage = markup;
    }

    // 4 Дополнительных/переопределённых метода
    public override void SubmitOrder(Product product, int quantity)
    {
        Console.WriteLine($"Покупатель оформил заказ в '{CompanyName}'. Доставка в магазин по адресу: {StoreLocation}.");
    }

    public void CalculateRetailPrice(Product product)
    {
        decimal finalPrice = product.Price * (1.0m + MarkupPercentage);
        Console.WriteLine($"Розничная цена '{product.Name}' в {CompanyName} (с наценкой {MarkupPercentage:P0}): {finalPrice:C}.");
    }

    public void OrderFromManufacturer(Manufacturer manufacturer, Product product, int quantity)
    {
        Console.WriteLine($"\n[B2B ЗАПРОС] Ритейлер '{CompanyName}' заказывает у завода '{manufacturer.CompanyName}'...");
        manufacturer.ProvideQuote(product);
        manufacturer.SubmitOrder(product, quantity);
        Console.WriteLine($"Товар поступит на склад: {StoreLocation}.");
    }

    public void LaunchPromotion(string promoName)
    {
        Console.WriteLine($"Магазин {CompanyName} запустил промо-акцию: '{promoName}'.");
    }
}

// Потомок Retailer -> демонстрация сложного (многоуровневого) наследования
public class ChainStore : Retailer
{
    public string RegionalHeadquarters { get; set; }

    public ChainStore(int supplierId, string companyName, string productType, double rating,
                      string storeLocation, int outlets, bool onlineShop, decimal markup, string headquarters)
        : base(supplierId, companyName, productType, rating, storeLocation, outlets, onlineShop, markup)
    {
        RegionalHeadquarters = headquarters;
    }

    public void ManageFranchise()
    {
        Console.WriteLine($"Сеть '{CompanyName}' управляет {RetailOutletsCount} филиалами из штаб-квартиры в г. {RegionalHeadquarters}.");
    }
}

public class Importer : Supplier, ICustomsClearance, ILogistics
{
    // 4 Дополнительных атрибута
    public string OriginCountry { get; set; }
    public string DeclarationNumber { get; set; }
    public string TransportType { get; set; }
    public decimal ImportTaxRate { get; set; }

    public Importer(int supplierId, string companyName, string productType, double rating,
                    string originCountry, string declarationNumber, string transportType, decimal taxRate)
        : base(supplierId, companyName, productType, rating)
    {
        OriginCountry = originCountry;
        DeclarationNumber = declarationNumber;
        TransportType = transportType;
        ImportTaxRate = taxRate;
    }

    // 4 Дополнительных/переопределённых метода
    public override string GetCompanyInfo()
    {
        return $"{base.GetCompanyInfo()} | Страна импорта: {OriginCountry} | Доставка: {TransportType}";
    }

    public bool ClearCustoms()
    {
        Console.WriteLine($"Таможенная декларация №{DeclarationNumber} одобрена. Пошлина {ImportTaxRate:P0} уплачена.");
        return true;
    }

    public void CalculateDeliveryTime(double distanceKm)
    {
        double hours = distanceKm / 60.0; // Примерная скорость
        Console.WriteLine($"Расчетное время доставки ({TransportType}, {distanceKm} км): ~{hours:F1} ч.");
    }

    public void CalculateLandDuty(Product product)
    {
        decimal taxAmount = product.Price * ImportTaxRate;
        Console.WriteLine($"Таможенный сбор на '{product.Name}' из страны {OriginCountry}: {taxAmount:C}.");
    }
}


Product machine = new Product("Промышленный станок ЧПУ", 500000m);
Product phone = new Product("Смартфон", 45000m);

Manufacturer factory = new Manufacturer(1, "ТехноПром", "Промышленные станки", 4.9, 1995, 1200, true, 0.20m);
ChainStore retailChain = new ChainStore(2, "МегаМаркет", "Бытовая электроника", 4.5, "г. Екатеринбург, ул. Ленина, 50", 45, true, 0.15m, "Екатеринбург");
Importer importer = new Importer(3, "ГлобалТрейд", "Экзотические фрукты", 4.2, "Коста-Рика", "DEC-109283", "Морской фрахт", 0.12m);

List<Supplier> suppliers = new List<Supplier> { factory, retailChain, importer };

Console.WriteLine("=== 1. ИНФОРМАЦИЯ О ПОСТАВЩИКАХ (ПОЛИМОРФИЗМ) ===");
foreach (var supplier in suppliers)
{
    Console.WriteLine(supplier.GetCompanyInfo());
}

Console.WriteLine("\n=== 2. ПРОСТОЕ НАСЛЕДОВАНИЕ (Manufacturer) ===");
factory.ProvideQuote(machine);
factory.InspectQuality();
factory.PlanProductionRun(50);

Console.WriteLine("\n=== 3. СЛОЖНОЕ / МНОГОУРОВНЕВОЕ НАСЛЕДОВАНИЕ (ChainStore : Retailer : Supplier) ===");
retailChain.CalculateRetailPrice(phone);
retailChain.ManageFranchise();
retailChain.OrderFromManufacturer(factory, machine, 5);

Console.WriteLine("\n=== 4. МНОЖЕСТВЕННОЕ НАСЛЕДОВАНИЕ ЧЕРЕЗ ИНТЕРФЕЙСЫ (Importer) ===");
importer.ClearCustoms();
importer.CalculateDeliveryTime(2400);
importer.CalculateLandDuty(phone);
```
