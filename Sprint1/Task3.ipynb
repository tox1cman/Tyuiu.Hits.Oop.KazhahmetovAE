### Вариант задания №14


<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Описание задачи:
Создать базовый класс Supplier в C#, который будет представлять информацию о
поставщиках товаров или услуг. На основе этого класса разработать 2-3
производных класса, демонстрирующих принципы наследования и полиморфизма.
В каждом из классов должны быть реализованы новые атрибуты и методы, а также
переопределены некоторые методы базового класса для демонстрации
полиморфизма.
Требования к базовому классу Supplier:
• Атрибуты: ID поставщика (SupplierId), Название компании (CompanyName),
Тип продукции (ProductType).
• Методы:
o GetCompanyInfo(): метод для получения информации о компании.
o ProvideQuote(): метод для предоставления котировки на товары или
услуги.
o SubmitOrder(): метод для отправки заказа поставщику.
Требования к производным классам:
1. Производитель (Manufacturer): Должен содержать дополнительные атрибуты,
такие как Год основания (FoundedYear). Метод ProvideQuote() должен быть
переопределен для включения информации о годе основания компании в
котировку.
2. Ритейлер (Retailer): Должен содержать дополнительные атрибуты, такие как
Расположение магазина (StoreLocation). Метод SubmitOrder() должен быть
переопределен для добавления информации о расположении магазина при
отправке заказа.
3. Импортер (Importer) (если требуется третий класс): Должен содержать
дополнительные атрибуты, такие как Страна происхождения товара
(OriginCountry). Метод GetCompanyInfo() должен быть переопределен для
отображения страны происхождения товара вместе с остальной
информацией о компании.

#### Дополнительное задание
Добавьте к сущестующим классам конструктора классов с использованием гетторов и сетторов и реализуйте взаимодействие объектов между собой


<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;
using System.Collections.Generic;



public class Product
{
    private string _name;
    private decimal _price;

    public string Name
    {
        get => _name;
        set => _name = string.IsNullOrWhiteSpace(value) ? "Неизвестный товар" : value;
    }

    public decimal Price
    {
        get => _price;
        set => _price = value < 0 ? 0 : value;
    }

    public Product(string name, decimal price)
    {
        Name = name;
        Price = price;
    }
}

public class Supplier
{
    private int _supplierId;
    private string _companyName;
    private string _productType;

    public int SupplierId
    {
        get => _supplierId;
        set => _supplierId = value < 0 ? 0 : value;
    }

    public string CompanyName
    {
        get => _companyName;
        set => _companyName = string.IsNullOrWhiteSpace(value) ? "Без названия" : value;
    }

    public string ProductType
    {
        get => _productType;
        set => _productType = string.IsNullOrWhiteSpace(value) ? "Разное" : value;
    }

    public Supplier(int supplierId, string companyName, string productType)
    {
        SupplierId = supplierId;
        CompanyName = companyName;
        ProductType = productType;
    }

    public virtual string GetCompanyInfo()
    {
        return $"[ID: {SupplierId}] Компания: {CompanyName} | Продукция: {ProductType}";
    }

    public virtual void ProvideQuote(Product product)
    {
        Console.WriteLine($"Котировка от {CompanyName}: базовое предложение на {product.Name} по цене {product.Price:C}.");
    }

    public virtual void SubmitOrder(Product product, int quantity)
    {
        Console.WriteLine($"Заказ на {quantity} шт. товара '{product.Name}' отправлен поставщику {CompanyName}.");
    }
}

public class Manufacturer : Supplier
{
    private int _foundedYear;

    public int FoundedYear
    {
        get => _foundedYear;
        set => _foundedYear = (value < 1800 || value > DateTime.Now.Year) ? 2000 : value;
    }

    public Manufacturer(int supplierId, string companyName, string productType, int foundedYear)
        : base(supplierId, companyName, productType)
    {
        FoundedYear = foundedYear;
    }

    public override void ProvideQuote(Product product)
    {
        decimal wholesalePrice = product.Price * 0.8m;
        Console.WriteLine($"Завод {CompanyName} (осн. {FoundedYear} г.): Оптовая цена на '{product.Name}' — {wholesalePrice:C}.");
    }
}

public class Retailer : Supplier
{
    private string _storeLocation;

    public string StoreLocation
    {
        get => _storeLocation;
        set => _storeLocation = string.IsNullOrWhiteSpace(value) ? "Адрес не указан" : value;
    }

    public Retailer(int supplierId, string companyName, string productType, string storeLocation)
        : base(supplierId, companyName, productType)
    {
        StoreLocation = storeLocation;
    }

    public void OrderFromManufacturer(Manufacturer manufacturer, Product product, int quantity)
    {
        Console.WriteLine($"\n[ВЗАИМОДЕЙСТВИЕ] Ритейлер '{CompanyName}' делает запрос производителю '{manufacturer.CompanyName}'...");
        manufacturer.ProvideQuote(product);
        manufacturer.SubmitOrder(product, quantity);
        Console.WriteLine($"Товар '{product.Name}' в количестве {quantity} шт. поступает на склад по адресу: {StoreLocation}.");
    }

    public override void SubmitOrder(Product product, int quantity)
    {
        Console.WriteLine($"Покупатель оформил заказ в '{CompanyName}'. Доставка в магазин по адресу: {StoreLocation}.");
    }
}

public class Importer : Supplier
{
    private string _originCountry;

    public string OriginCountry
    {
        get => _originCountry;
        set => _originCountry = string.IsNullOrWhiteSpace(value) ? "Неизвестна" : value;
    }

    public Importer(int supplierId, string companyName, string productType, string originCountry)
        : base(supplierId, companyName, productType)
    {
        OriginCountry = originCountry;
    }

    public override string GetCompanyInfo()
    {
        return $"{base.GetCompanyInfo()} | Страна происхождения: {OriginCountry}";
    }
}



Product machine = new Product("Промышленный станок ЧПУ", 500000m);
Product phone = new Product("Смартфон", 45000m);

Manufacturer factory = new Manufacturer(1, "ТехноПром", "Промышленные станки", 1995);
Retailer store = new Retailer(2, "МегаМаркет", "Бытовая электроника", "г. Екатеринбург, ул. Ленина, 50");
Importer importer = new Importer(3, "ГлобалТрейд", "Экзотические фрукты", "Коста-Рика");

List<Supplier> suppliers = new List<Supplier> { factory, store, importer };

Console.WriteLine("=== ИНФОРМАЦИЯ О ПОСТАВЩИКАХ ===");
foreach (var supplier in suppliers)
{
    Console.WriteLine(supplier.GetCompanyInfo());
}

Console.WriteLine("\n=== ПРЕДОСТАВЛЕНИЕ КОТИРОВОК ===");
factory.ProvideQuote(machine);
store.ProvideQuote(phone);

Console.WriteLine("\n=== ВЗАИМОДЕЙСТВИЕ ОБЪЕКТОВ МЕЖДУ СОБОЙ ===");
store.OrderFromManufacturer(factory, machine, 5);
```
